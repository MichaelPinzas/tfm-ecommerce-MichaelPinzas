# Exploración Inicial del Dataset E-commerce

**Dataset:** eCommerce behavior data from Multi Category Store (Kaggle)

**Archivos:**
- 2019-Oct.csv (5.4 GB)
- 2019-Nov.csv (8.6 GB)

**Objetivo:** Entender la estructura, tipos de datos y contenido del dataset.

```python
import pandas as pd
import sys
from pathlib import Path

# Añadir src al path
project_root = Path.cwd().parent
sys.path.insert(0, str(project_root))

from src.utils import RAW_DATA_DIR, setup_logger

logger = setup_logger(__name__)
logger.info("Notebook iniciado")
```

```python
# Cargar solo las primeras 100,000 filas para exploración rápida
file_path = RAW_DATA_DIR / "2019-Oct.csv"

logger.info(f"Cargando muestra de {file_path.name}...")

# Leer solo primeras 100k filas
df_sample = pd.read_csv(file_path, nrows=100000)

logger.info(f"Muestra cargada: {len(df_sample):,} filas")
df_sample.head()
```

```python
# Información general del dataset
print("=" * 50)
print("INFORMACIÓN DEL DATASET")
print("=" * 50)
df_sample.info()
```

```python
# Estadísticas de columnas numéricas
print("\n" + "=" * 50)
print("ESTADÍSTICAS BÁSICAS")
print("=" * 50)
df_sample.describe()
```

```python
# Distribución de tipos de eventos
print("\n" + "=" * 50)
print("DISTRIBUCIÓN DE TIPOS DE EVENTOS")
print("=" * 50)
event_counts = df_sample['event_type'].value_counts()
print(event_counts)
print(f"\nPorcentajes:")
print(event_counts / len(df_sample) * 100)
```

```python
# Valores nulos por columna
print("\n" + "=" * 50)
print("VALORES NULOS")
print("=" * 50)
null_counts = df_sample.isnull().sum()
null_percentages = (null_counts / len(df_sample) * 100).round(2)
null_df = pd.DataFrame({
    'Nulos': null_counts,
    'Porcentaje': null_percentages
})
print(null_df[null_df['Nulos'] > 0])
```

```python
# Top 10 categorías más vistas
print("\n" + "=" * 50)
print("TOP 10 CATEGORÍAS")
print("=" * 50)
top_categories = df_sample['category_code'].value_counts().head(10)
print(top_categories)
```
